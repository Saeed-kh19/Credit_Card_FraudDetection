# Credit Card Fraud Detection: Library-Based Outlier Detection

**Dataset:** [Kaggle — Credit Card Fraud Detection](https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud?resource=download)

**Goal:** Apply multiple outlier-detection methods to numeric transaction features, evaluate their predictions against the labelled `Class` column, compare the detected rows, and find the exact rows shared by all methods.

We use six methods: Mahalanobis Distance, Isolation Forest, Local Outlier Factor (LOF), DBSCAN, PCA Reconstruction Error, and k-Nearest-Neighbour (kNN) Distance.

> **Library-first implementation:** The algorithms are implemented by scikit-learn estimators and their built-in methods. We do not manually code the internal algorithm mathematics. Our own code handles data preparation, score thresholding, evaluation, and row-ID intersections.

> **Important:** `Class` is not an input feature and is never passed to a detector. It is kept separately for stratified benchmark sampling and evaluation. The known class prevalence is used only as a common score-based alert-budget prior, as explained in the methodology.

## Step 0 — Environment and configuration

Install the packages from `requirements.txt` before running this notebook. Put the downloaded `creditcard.csv` file in the project's `data/` directory. The default benchmark uses 30,000 stratified rows so LOF and DBSCAN remain more practical on a typical computer. All methods process exactly the same rows.

The complete detector implementations are in `run_analysis.py` and use scikit-learn APIs such as `LedoitWolf.mahalanobis`, `IsolationForest.score_samples`, `LocalOutlierFactor.fit_predict`, `DBSCAN.fit_predict`, `PCA.inverse_transform`, and `NearestNeighbors.kneighbors`.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display

from run_analysis import (
    load_and_validate,
    select_benchmark_sample,
    preprocess_features,
    run_detectors,
    calculate_metrics,
    calculate_overlaps,
    export_results,
)

DATA_PATH = Path("data/creditcard.csv")
OUTPUT_DIR = Path("outputs")
MAX_ROWS = 30_000       # Set to 0 to use the full dataset (LOF/DBSCAN may be slow).
DBSCAN_EPS_QUANTILE = 0.98
RANDOM_STATE = 42

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

## Step 1 — Load the dataset and inspect its structure

The dataset contains 30 numeric input features: `Time`, `V1`–`V28`, and `Amount`. `Class` is the binary label (`0` = legitimate transaction, `1` = fraudulent transaction). We preserve a zero-based `Transaction_ID` based on the original CSV row position so detected outliers can be traced back to the original file.

We inspect missing values, duplicated rows, and class counts. Exact duplicates are counted but not automatically removed because repeated transactions may be meaningful observations.

```python
df, feature_cols = load_and_validate(DATA_PATH)

dataset_rows = len(df)
full_data_fraud_rate = float(df["Class"].mean())

print("Dataset shape:", df.shape)
print("Feature count:", len(feature_cols))
print("Full-data fraud rate:", f"{full_data_fraud_rate:.4%}")
print("Class distribution:")
display(df["Class"].value_counts().rename(index={0: "Legitimate", 1: "Fraud"}).to_frame("Count"))
```

## Step 2 — Create a common benchmark sample

Running LOF and DBSCAN on all 284,807 rows can be expensive. The default notebook therefore creates a stratified random sample of 30,000 rows. Stratification preserves the original class ratio and makes it more likely that the rare fraud class is represented adequately.

The labels are used to preserve the sample's class proportion, but they are not passed into any detector. Every method is run on exactly the same sampled transactions. Set `MAX_ROWS = 0` to use every row.

```python
sample = select_benchmark_sample(df, MAX_ROWS)

print("Benchmark shape:", sample.shape)
print("Benchmark fraud rate:", f"{sample['Class'].mean():.4%}")
display(sample[feature_cols + ["Class"]].head())
```

## Step 3 — Preprocess the numeric features

Preprocessing is important because Mahalanobis Distance, LOF, DBSCAN, and kNN rely on distances or local density.

1. Separate the 30 numeric input features from `Class`.
2. Apply `log1p` to `Amount` to reduce its right-skew.
3. Use median imputation only if a feature contains missing values.
4. Standardize the input features with `StandardScaler`.

The resulting model matrix contains only numeric features and no label column.

```python
X = preprocess_features(sample, feature_cols)
y = sample["Class"].astype(int).to_numpy()  # Ground truth used only after detection
transaction_ids = sample["Transaction_ID"].astype(int).to_numpy()

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Fraud cases in benchmark:", int(y.sum()))
```

## Step 4 — Run the outlier-detection methods using scikit-learn

The following library classes perform the actual algorithms; no algorithm internals are reimplemented manually:

| Method | Library API used in `run_analysis.py` |
|---|---|
| Mahalanobis Distance | `LedoitWolf().fit(X).mahalanobis(X)` |
| Isolation Forest | `IsolationForest().fit(X)` and `score_samples(X)` |
| Local Outlier Factor | `LocalOutlierFactor().fit_predict(X)` and `negative_outlier_factor_` |
| DBSCAN | `DBSCAN().fit_predict(X)`; label `-1` means noise |
| PCA Reconstruction | `PCA().fit_transform(X)` and `inverse_transform()` |
| kNN Distance | `NearestNeighbors().fit(X)` and `kneighbors(X)` |

DBSCAN and kNN use a 10-component PCA representation to reduce the dimensionality before density and neighbourhood calculations. PCA Reconstruction Error uses a separate PCA model retaining at least 95% of the variance.

For the five methods that produce continuous scores, the project flags the top `k` anomaly scores, where `k` is based on the labelled dataset's fraud prevalence. This is a common alert-budget threshold for comparison, not a feature used by the algorithms. DBSCAN uses its native noise labels and may flag a different number of rows.

The reusable function below fits all six library implementations and returns their scores, binary predictions, and parameter metadata.

```python
scores, predictions, metadata = run_detectors(
    X=X,
    transaction_ids=transaction_ids,
    fraud_rate=full_data_fraud_rate,
    dbscan_eps_quantile=DBSCAN_EPS_QUANTILE,
)

print("Methods:", list(predictions.keys()))
print("Top-k threshold budget:", metadata["expected_k"])
print("DBSCAN eps:", metadata["dbscan_eps"])
print("DBSCAN noise points:", metadata["dbscan_noise_count"])
```

## Step 5 — Quantitatively evaluate the methods

The `Class` column is now used as ground truth. For every method we calculate:

- **TP:** fraud transactions correctly flagged.
- **FP:** legitimate transactions incorrectly flagged.
- **FN:** fraud transactions missed by the detector.
- **TN:** legitimate transactions correctly left unflagged.
- **Precision:** proportion of flagged transactions that are fraud.
- **Recall:** proportion of all fraud transactions that are detected.
- **F1:** harmonic mean of precision and recall.
- **Specificity** and **balanced accuracy:** helpful under severe class imbalance.
- **ROC-AUC** and **Average Precision:** computed from continuous scores where a method provides them. DBSCAN has no native continuous score, so these two columns are reported as not applicable for DBSCAN.

Accuracy is not the main metric because legitimate transactions vastly outnumber fraudulent transactions.

```python
metrics = calculate_metrics(y, predictions, scores)
display(metrics.round(4))
```

## Step 6 — Compare detected rows and calculate intersections

Each detector returns a set of `Transaction_ID` values. We calculate:

1. **Pairwise intersection:** how many rows are shared by each pair of methods.
2. **Jaccard similarity:** intersection size divided by union size for each pair.
3. **All-method intersection:** rows flagged by every detector. This is the strictest definition of common outliers.
4. **3-or-more-method consensus:** rows flagged by at least three methods. This is useful when the strict all-method intersection is empty or very small.

The intersection is based on original CSV row positions, not the temporary row numbers of the sampled DataFrame.

```python
overlap_counts, consensus, common_all_ids, common = calculate_overlaps(
    transaction_ids,
    predictions,
    y_true=y,
)
jaccard = overlap_counts.attrs["jaccard"]

print("Pairwise intersection counts:")
display(overlap_counts)
print("Pairwise Jaccard similarities:")
display(jaccard.round(4))
print(f"Rows flagged by all {len(predictions)} methods: {len(common)}")
print("Common Transaction_ID values:", sorted(common_all_ids))
print("Rows flagged by at least 3 methods:", len(consensus))

display(common.head(50))
display(consensus.head(20))
```

## Step 7 — Export the results, plots, and report

This final step saves the complete row-level predictions, per-method outlier lists, confusion matrices, PR curves, pairwise overlap tables, the exact all-method intersection, a 3-or-more-method consensus set, and a Markdown report containing the measured results. The report is generated from this run; do not manually invent or fill in metrics.

```python
export_results(
    output_dir=OUTPUT_DIR,
    data_path=DATA_PATH,
    dataset_rows=dataset_rows,
    fraud_rate=full_data_fraud_rate,
    sample=sample,
    feature_columns=feature_cols,
    y=y,
    transaction_ids=transaction_ids,
    scores=scores,
    predictions=predictions,
    metrics=metrics,
    overlap_counts=overlap_counts,
    consensus=consensus,
    common_ids=common_all_ids,
    common=common,
    metadata=metadata,
)

print("Saved outputs to:", OUTPUT_DIR.resolve())
print("Main report:", OUTPUT_DIR / "results_report.md")
print("All-method intersection:", OUTPUT_DIR / "common_outliers_all_methods.csv")

# Display the most useful final output tables.
display(metrics.round(4))
display(common)
display(consensus.head(30))
```

## Step 8 — How to interpret the findings

Use the generated `outputs/results_report.md` and the CSV files to write the final discussion.

- A method with high recall detects more fraud cases, but may also flag many legitimate transactions.
- A method with high precision produces a cleaner alert list, but may miss more fraud.
- Average Precision is particularly useful because it summarizes the precision–recall trade-off across thresholds.
- DBSCAN's output depends on `eps`, `min_samples`, and the PCA representation. You should mention this sensitivity and, ideally, compare a second value such as `DBSCAN_EPS_QUANTILE = 0.99`.
- A strict all-method intersection may be empty. That is a valid result: it means no row was simultaneously flagged by every method under the chosen settings. Do not force the intersection to contain rows.
- Fraud is not exactly the same as statistical outlier: a fraud transaction may look statistically ordinary, while a legitimate transaction may be unusual.

**Reproducibility:** keep the random seed fixed, report `MAX_ROWS`, and include the DBSCAN parameters and the generated `run_summary.json` with your submission.
